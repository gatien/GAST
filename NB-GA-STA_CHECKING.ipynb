### Affichage des multivecteurs en notation centrale

```python
def display_ga(mv):
    """
    Affiche un multivecteur avec les substitutions de verseurs personnalisées.
    Version robuste : travaille sur les objets sympy/galgebra plutôt que sur
    la représentation en chaîne de caractères.
    """

    ga = mv.Ga

    basis_mapping = {
        'e_2*e_3': (1, r'\mathbf{B}_1'),
        'e_1*e_3': (-1, r'\mathbf{B}_2'),
        'e_1*e_2': (1, r'\mathbf{B}_3'),
        'e_1*e_2*e_3': (1, r'\mathscr{I}_\mathbb{1}'),
        'e_0*e_1*e_2*e_3': (1, r'\mathscr{I}_\mathbb{2}'),
        'e_0': (1, r'\mathscr{I}_\mathbb{3}'),
        'e_1': (-1, r'\mathscr{I}_\mathbb{1} \mathbf{B}_1'),
        'e_2': (-1, r'\mathscr{I}_\mathbb{1} \mathbf{B}_2'),
        'e_3': (-1, r'\mathscr{I}_\mathbb{1} \mathbf{B}_3'),
        'e_0*e_2*e_3': (1, r'\mathscr{I}_\mathbb{3} \mathbf{B}_1'),
        'e_0*e_1*e_3': (-1, r'\mathscr{I}_\mathbb{3} \mathbf{B}_2'),
        'e_0*e_1*e_2': (1, r'\mathscr{I}_\mathbb{3} \mathbf{B}_3'),
        'e_0*e_1': (-1, r'\mathscr{I}_\mathbb{2} \mathbf{B}_1'),
        'e_0*e_2': (-1, r'\mathscr{I}_\mathbb{2} \mathbf{B}_2'),
        'e_0*e_3': (-1, r'\mathscr{I}_\mathbb{2} \mathbf{B}_3'),
    }

    result_terms = []

    # 1. Partie scalaire, gérée séparément
    scalar_part = simplify(mv.scalar())
    if scalar_part != 0:
        result_terms.append(latex(scalar_part))

    # 2. Blades sous forme d'objets Mv
    blades = list(ga.mv_blades.flat[1:])
    coefs = mv.blade_coefs(blades)

    for blade, coeff in zip(blades, coefs):
        coeff = simplify(coeff)
        if coeff == 0:
            continue

        # >>> Correction : normaliser la notation wedge '^' en '*'
        # pour que la clé corresponde à celles du basis_mapping,
        # qui ont été définies avec la notation produit géométrique.
        key = str(blade).replace('^', '*')

        key_sign, symbol = basis_mapping.get(key, (1, None))
        if symbol is None:
            symbol = latex(blade)  # fallback si le blade n'est pas dans le mapping

        final_coeff = coeff * key_sign
        coeff_latex = latex(final_coeff)

        if final_coeff == 1:
            term = symbol
        elif final_coeff == -1:
            term = f"-{symbol}"
        elif final_coeff.is_Add:
            term = rf"\left({coeff_latex}\right) {symbol}"
        else:
            term = f"{coeff_latex} {symbol}"

        result_terms.append(term)

    if not result_terms:
        result = '0'
    else:
        result = result_terms[0]
        for term in result_terms[1:]:
            result += f" {term}" if term.startswith('-') else f" + {term}"

    display(Math(result))
    #return result
    return
    
# Tests
print("\n=== Tests de display_ga ===")

print("\nTest 1: e1")
display_ga(e1)

print("\nTest 2: e0")
display_ga(e0)

print("\nTest 3: e2^e3")
display_ga(e2^e3)

print("\nTest 4: 2*e1 + 3*(e2^e3)")
display_ga(2*e1 + 3*(e2^e3))

print("\nTest 5: e1^e2^e3")
display_ga(e1^e2^e3)

print("\nTest 6: -e1 + e0")
display_ga(-e1 + e0)
```

### Fonctions pour la vérification des constantes de struture

On définit d'abord une fonction *expand_su3_structure_constants* pour étendre le dictionnaire contenant les valeurs des constantes de struture.

La fonction *verify_su3_structure_constants* détermine si les commutateurs des opérateurs appliqués sur un mv (arbitraire mais fixe et bien choisi ;-) ) donne le même résultat que celui calculé à partir de l'action des opérateurs déterminés par les constantes de structure. Attention de bien choisir le *mv_test* et le *nb_complex*

```python
def expand_su3_structure_constants(f_constants_partial):
    """
    Expande les constantes de structure SU(3) en ajoutant toutes les permutations
    antisymétriques des indices.
    
    Pour chaque triplet (i, j, k) avec valeur v :
    - Les permutations cycliques (j, k, i) et (k, i, j) ont la même valeur v
    - Les permutations non-cycliques ont la valeur opposée -v
    
    Args:
        f_constants_partial: Dictionnaire partiel avec quelques constantes de base
    
    Returns:
        f_constants: Dictionnaire complet avec toutes les permutations
    """
    f_constants = f_constants_partial.copy()
    
    # Parcourir une copie des éléments pour ne pas modifier pendant l'itération
    for (i, j, k), val in list(f_constants_partial.items()):
        # Permutations cycliques : même valeur
        f_constants[(j, k, i)] = val
        f_constants[(k, i, j)] = val
        
        # Permutations non-cycliques : valeur opposée
        f_constants[(j, i, k)] = -val
        f_constants[(i, k, j)] = -val
        f_constants[(k, j, i)] = -val
    
    return f_constants


# Votre dictionnaire de base
f_constants_partial = {
    # f^1_23 = 1
    (1, 2, 3): 1,
    (4, 5, 8): sqrt(3)/2,
    (6, 7, 8): sqrt(3)/2,
    (1, 4, 7): Rational(1, 2),
    (2, 4, 6): Rational(1, 2),
    (2, 5, 7): Rational(1, 2),
    (3, 4, 5): Rational(1, 2),
    (1, 5, 6): Rational(-1, 2),
    (3, 6, 7): Rational(-1, 2),
}

# Expansion complète
f_constants = expand_su3_structure_constants(f_constants_partial)

## Affichage du résultat
#print(f"Constantes de base: {len(f_constants_partial)} éléments")
#print(f"Constantes après expansion: {len(f_constants)} éléments\n")
#
## Optionnel : affichage trié des constantes non nulles
#print("Constantes de structure (f_{abc} non nulles):")
#for (a, b, c) in sorted(f_constants.keys()):
#    val = f_constants[(a, b, c)]
#    if val != 0:
#        print(f"  f_{a}{b}{c} = {val}")

f_constants
```

```python

def verify_su3_structure_constants(λList, mv_test, f_constants,nb_complex,verbose=False):
    """
    Vérifie les constantes de structure de SU(3) pour une liste d'opérateurs (fonctions).
    
    Args:
        λList: Liste des 8 opérateurs (fonctions qui prennent un Mv et retournent un Mv)
               Les indices sont 0-7, correspondant aux λ_1 à λ_8
        mv_test: Un multivecteur test pour évaluer les commutateurs
        f_constants: Dictionnaire des constantes de structure f^a_bc.
                    Clés: tuples (a, b, c) avec a, b, c numérotés de 1 à 8
                    Si None, utilise les valeurs standard de SU(3)
    
    Returns:
        Un dictionnaire avec les résultats de vérification
    """
    
    results = {}
    num_operators = len(λList)
    
    print(f"\n{'='*80}")
    print(f"Vérification des constantes de structure SU(3)")
    print(f"{'='*80}\n")
    
    passed = 0
    failed = 0
    
    # Boucle sur les indices de la liste (0-7)
    for idx_a in range(num_operators):
        for idx_b in range(idx_a+1, num_operators):
            # Conversion des indices de liste (0-7) vers la numérotation des constantes (1-8)
            a = idx_a + 1
            b = idx_b + 1
            
            # Calcul du commutateur [λ_a, λ_b] appliqué sur mv_test
            # Les opérateurs sont des fonctions, donc on les appelle
            λ_a_mv = λList[idx_a](mv_test)
            λ_b_mv = λList[idx_b](mv_test)
            
            # Commutateur : [λ_a, λ_b](v) = λ_a(λ_b(v)) - λ_b(λ_a(v))
            commutator_result = λList[idx_a](λList[idx_b](mv_test)) - λList[idx_b](λList[idx_a](mv_test))
            
            # Calcul du côté droit : Σ_c f^c_ab * λ_c(v)
            structure_result = None
            
            for idx_c in range(num_operators):
                # Conversion de l'indice c (0-7) vers la numérotation (1-8)
                c = idx_c + 1
                
                f_cab = f_constants.get((c, a, b), 0)
                
                if f_cab != 0:
                    # Si nombre complex agit à droite
                    term = 2*f_cab * λList[idx_c](mv_test)*nb_complex

                    # Si nombre complex agit à gauche:
                    # term = 2*f_cab * nb_complex * λList[idx_c](mv_test)
                    
                    if structure_result is None:
                        structure_result = term
                    else:
                        structure_result = structure_result + term
            
            if structure_result is None:
                structure_result = 0 * mv_test  # Cas où tous les f^c_ab = 0
            
            # Comparaison
            difference = commutator_result - structure_result
            
            # Simplification et vérification
            try:
                diff_simplified = difference.simplify()
                is_zero = diff_simplified == 0
            except:
                is_zero = False
            
            key = f"[λ_{a}, λ_{b}]"
            results[key] = {
                'is_valid': is_zero,
                'commutator': commutator_result,
                'structure_result': structure_result,
                'difference': difference,
                'indices_list': (idx_a, idx_b)
            }
            
            if is_zero:
                if verbose:
                    print(f"✓ {key}: VALIDE")
                passed += 1
            else:
                if verbose:
                    print(f"✗ {key}: INVALIDE")
                    print(f"  Différence non nulle")
                failed += 1
    
            if is_zero:
                if verbose:
                    display(Math(f"\\checkmark \\quad [\\lambda_{{{a}}}, \\lambda_{{{b}}}]: \\textbf{{VALIDE}}"))
                passed += 1
            else:
                if verbose:
                    display(Math(f"\\times \\quad [\\lambda_{{{a}}}, \\lambda_{{{b}}}]: \\textbf{{INVALIDE}}"))
                failed += 1
    
    display(Math(""))
    display(Math(f"\\textbf{{Résumé: {passed} validé(s), {failed} échoué(s) sur {passed+failed} relation(s)}}"))
   
    return results


def display_verification_details(results, key):
    """
    Affiche les détails d'une vérification spécifique
    """
    if key not in results:
        display(Math(f"\\text{{Clé '{key}' non trouvée}}"))
        return
    result = results[key]
    display(Math(f"\\textbf{{Détails pour {key}}}"))
    display(Math("\\text{Commutateur appliqué:}"))
    display_ga(result['commutator'])
    display(Math("\\text{Résultat de la structure:}"))
    display_ga(result['structure_result'])
    display(Math("\\text{Différence:}"))
    display_ga(result['difference'])
    status = "✓ VALIDE" if result['is_valid'] else "✗ INVALIDE"
    display(Math(f"\\text{{{status}}}"))
```

## Self-adjoint operators checking
We test below that $\lambda_a(\mathcal{M}) * \mathcal{O} = \mathcal{M} * \lambda_a(\mathcal{O})$ for arbitrary mv $\mathcal{M}$ and $\mathcal{O}$ (in the case here of bivector SU(3) in fundamental representation)

```python
def AdjointTest(opList,Mc,Oc,verbose=False):
    results = {}
    print(f"\n{'='*80}")
    print(f"Vérification du caractère autoadjoint des générateurs pour SU(3)")
    print(f"{'='*80}\n")
    
    passed = 0
    failed = 0
    for i,op in enumerate(opList):
        # calculation of scalar products
        a=i+1
        ScalProdOpLeft=scalprod(op(Mc),Oc)
        ScalProdOpRigt=scalprod(Mc,op(Oc))
        
        # Comparaison
        difference = ScalProdOpLeft - ScalProdOpRigt
        
        # Simplification et vérification
        try:
            diff_simplified = difference.simplify()
            is_zero = diff_simplified == 0
        except:
            is_zero = False
        
        key = f"λ_{a}"
        results[key] = {
            'is_valid': is_zero,
            'left': ScalProdOpLeft,
            'right': ScalProdOpRigt,
            'difference': difference
        }
        
        if is_zero:
            if verbose:
                print(f"✓ {key}: VALIDE")
            passed += 1
        else:
            if verbose:
                print(f"✗ {key}: INVALIDE")
                print(f"  Différence non nulle")
            failed += 1
    
    display(Math(""))
    display(Math(f"\\textbf{{Résumé: {passed} validé(s), {failed} échoué(s) sur {passed+failed} relation(s)}}"))
   
    return results

def display_verification_details_adjoint(results, key):
    """
    Affiche les détails d'une vérification spécifique
    """
    if key not in results:
        display(Math(f"\\text{{Clé '{key}' non trouvée}}"))
        return
    result = results[key]
    display(Math(f"\\textbf{{Détails pour {key}}}"))
    display(Math("\\text{op appliqué à gauche:}"))
    display_ga(result['left'])
    display(Math("\\text{op appliqué à droite:}"))
    display_ga(result['right'])
    display(Math("\\text{Différence:}"))
    display_ga(result['difference'])
    status = "✓ VALIDE" if result['is_valid'] else "✗ INVALIDE"
    display(Math(f"\\text{{{status}}}"))
```

```python
# for fundamental SU(3) isoversor
adj_result=AdjointTest(λλList,Mc,Oc,verbose=False)
#display_verification_details_adjoint(adj_result, 'λ_4')

# for conjugate SU(3) isoversor
adj_result=AdjointTest(λλbList,Mc,Oc,verbose=False)

# for fundamental SU(3) bivector
adj_result=AdjointTest(λList,Mc,Oc,verbose=False)

# for conjugate SU(3) bivector
adj_result=AdjointTest(λbList,Mc,Oc,verbose=False)
```

## Raising and lowering operators 

### For fundamental and conjugate representation of isoversor and bivector SU(3)

By definition,$ \sigma =+1$ is for a raising operator and $\sigma = -1$ is for lowering operator

```python
# fund isoversor SU3
def Eαα0(σ,Mc):
    if σ==1 or σ==-1:
        return (-σ*λλ1(Mc)*B3+λλ2(Mc))/(2*sqrt(2))
def Eαα1(σ,Mc):
    if σ==1 or σ==-1:
        return (λλ4(Mc)+σ*λλ5(Mc)*B3)/(2*sqrt(2))
def Eαα2(σ,Mc):
    if σ==1 or σ==-1:
        return (-σ*λλ6(Mc)*B3-λλ7(Mc))/(2*sqrt(2))
EααList=[Eαα0,Eαα1,Eαα2]

# conj isoversor SU3
def Ebαα0(σ,Mc):
    if σ==1 or σ==-1:
        return (-σ*λλb1(Mc)*B3+λλb2(Mc))/(2*sqrt(2))
def Ebαα1(σ,Mc):
    if σ==1 or σ==-1:
        return (λλb4(Mc)+σ*λλb5(Mc)*B3)/(2*sqrt(2))
def Ebαα2(σ,Mc):
    if σ==1 or σ==-1:
        return (σ*λλb6(Mc)*B3+λλb7(Mc))/(2*sqrt(2))
EbααList=[Ebαα0,Ebαα1,Ebαα2]

# fund bivector SU3
def Eα0(σ,Mc):
    if σ==1 or σ==-1:
        return (λ1(Mc)+σ*λ2(Mc)*I2)/(2*sqrt(2))
def Eα1(σ,Mc):
    if σ==1 or σ==-1:
        return (λ4(Mc)+σ*λ5(Mc)*I2)/(2*sqrt(2))
def Eα2(σ,Mc):
    if σ==1 or σ==-1:
        return (λ6(Mc)-σ*λ7(Mc)*I2)/(2*sqrt(2))
EαList=[Eα0,Eα1,Eα2]

# conj bivector SU3
def Ebα0(σ,Mc):
    if σ==1 or σ==-1:
        return (λb1(Mc)-σ*λb2(Mc)*I2)/(2*sqrt(2))
def Ebα1(σ,Mc):
    if σ==1 or σ==-1:
        return (-λb4(Mc)-σ*λb5(Mc)*I2)/(2*sqrt(2))
def Ebα2(σ,Mc):
    if σ==1 or σ==-1:
        return (λb6(Mc)-σ*λb7(Mc)*I2)/(2*sqrt(2))
EbαList=[Ebα0,Ebα1,Ebα2]


# charge conjugation
def CC(Mc):
    return -Mc*I2*B1
```

### Different representations of $\gamma$ operators (chiral, standard, and others)

The three operators below change the representation. All commute pairwise.

```python
def U(σ,Mc):
    "U+ switches from chiral to std representation"
    return (Mc+σ*I1*Mc*I3*B3)/sqrt(2)
def V(σ,Mc):
    return (Mc+σ*I3*Mc*I2)/sqrt(2)
def W(σ,Mc):
    return (Mc+σ*I2*Mc*I1)/sqrt(2)
```

Below are examples of representation change for multivectors and operators

```python
# mv change of representation
def UoV(σ,Mc):
    return U(σ,V(σ,Mc))
def UoVoW(σ,Mc):
    return U(σ,V(σ,W(σ,Mc)))

# operator change of representation
def UoV_operators(Op):
    '''Change l'opérateur OP vers la représentation obtenue en composant U et V. 
    If the mv is transformed as well then the result should be UoV of the 
    transformed mv by the operator then changed of representation.'''
    return lambda Mc: U(+1,V(+1,Op(U(-1,V(-1,Mc)))))
def UoVoW_opeatores(Op):
    "Change l'opérateur OP vers la représentation obtenue en composant U,V et W"
    return lambda Mc: U(+1,V(+1,W(+1,Op(U(-1,V(-1,W(-1,Mc)))))))
```
